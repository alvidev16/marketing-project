### Evaluación de Normalidad para Variables Cuantitativas

```python
# Identificar columnas cuantitativas (numéricas)
quantitative_columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Excluir 'Unnamed: 0' y 'id' ya que no son variables para análisis estadístico de distribución
columns_to_exclude = ['Unnamed: 0', 'id']
quantitative_columns = [col for col in quantitative_columns if col not in columns_to_exclude]

print("Columnas cuantitativas para análisis de normalidad:", quantitative_columns)
```

```python
# Identificar columnas cuantitativas (numéricas)
quantitative_columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Excluir 'Unnamed: 0' y 'id' ya que no son variables para análisis estadístico de distribución
columns_to_exclude = ['Unnamed: 0', 'id']
quantitative_columns = [col for col in quantitative_columns if col not in columns_to_exclude]

for col in quantitative_columns:
    # Asegurarse de que no hay valores nulos en la columna actual para las pruebas
    current_data = df[col].dropna()

    # Si la columna está vacía después de eliminar nulos, saltar
    if current_data.empty:
        print(f"Saltando la columna '{col}' porque no tiene datos válidos después de eliminar nulos.")
        continue

    print(f"\n--- Análisis para la columna: {col} ---")

    plt.figure(figsize=(14, 6))

    # Histograma con Curva de Densidad de Kernel
    plt.subplot(1, 2, 1)
    sns.histplot(current_data, kde=True, color='#002D72')
    plt.title(f'Histograma y Densidad Sugerida ({col})')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')

    # Gráfico Q-Q para validación visual de normalidad
    plt.subplot(1, 2, 2)
    # stats.probplot returns a tuple of (osm, osr) and another tuple of (slope, intercept, r_value)
    (os_medians, os_res), (slope, intercept, r_value) = stats.probplot(current_data, dist="norm", plot=None)
    plt.plot(os_medians, os_res, 'o', color='#002D72', label='Data Points') # Data points
    plt.plot(os_medians, slope * os_medians + intercept, color='#002D72', linestyle='--', label='Theoretical Line') # Theoretical line
    plt.title(f'Gráfico Q-Q (Ajuste a la Diagonal) ({col})')
    plt.legend(loc='upper left') # Add legend to distinguish lines

    plt.tight_layout()
    plt.show()

    # Cálculo de estadísticos descriptivos de forma
    print(f"Sesgo (Skewness) para {col}:", stats.skew(current_data))
    print(f"Curtosis para {col}:", stats.kurtosis(current_data))

    # Ejecución de Pruebas de Bondad de Ajuste
    # H0: Los datos siguen una distribución normal
    # Shapiro-Wilk es más preciso para n <= 5000
    # Kolmogorov-Smirnov es más adecuado para muestras grandes o cuando no se asume una distribución específica
    if len(current_data) <= 5000:
        sw_stat, sw_p = stats.shapiro(current_data)
        print(f"Prueba Shapiro-Wilk para {col}: Estadístico={sw_stat:.4f}, p-valor={sw_p:.4e}")
    else:
        print(f"La prueba Shapiro-Wilk no es precisa para N > 5000 (N={len(current_data)}). Saltando esta prueba.")

    try:
        ks_stat, ks_p = stats.kstest(current_data, 'norm', args=(np.mean(current_data), np.std(current_data)))
        print(f"Prueba Kolmogorov-Smirnov para {col}: Estadístico={ks_stat:.4f}, p-valor={ks_p:.4e}")
    except Exception as e:
        print(f"Error al ejecutar Kolmogorov-Smirnov para {col}: {e}")
```

**Outliers**

```python
print("### Detección de Outliers utilizando los Cercos de Tukey (Tukey’s Fences) ###")

# Identificar columnas cuantitativas (numéricas) que ya se han definido previamente
# Se reutiliza la lista 'quantitative_columns' del análisis de normalidad

print("Columnas cuantitativas para la detección de outliers:", quantitative_columns)

# DataFrame para almacenar los outliers identificados y sus características
outliers_summary = pd.DataFrame(columns=['Variable', 'IQR', 'Lower Bound', 'Upper Bound', 'Number of Outliers', 'Percentage of Outliers', 'Outlier Indices'])

for col in quantitative_columns:
    current_data = df[col].dropna()

    if current_data.empty:
        print(f"Saltando la columna '{col}' porque no tiene datos válidos después de eliminar nulos.")
        continue

    # Calcular Q1 (Primer Cuartil), Q3 (Tercer Cuartil) y IQR (Rango Intercuartílico)
    Q1 = current_data.quantile(0.25)
    Q3 = current_data.quantile(0.75)
    IQR = Q3 - Q1

    # Calcular los límites de los Cercos de Tukey
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identificar outliers
    outliers = current_data[(current_data < lower_bound) | (current_data > upper_bound)]

    # Almacenar la información de los outliers
    outliers_summary = pd.concat([outliers_summary, pd.DataFrame([{
        'Variable': col,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': len(outliers),
        'Percentage of Outliers': (len(outliers) / len(current_data)) * 100 if len(current_data) > 0 else 0,
        'Outlier Indices': outliers.index.tolist()
    }])], ignore_index=True)

    print(f"\n--- Análisis de Outliers para la columna: {col} ---")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Límite Inferior (Lower Bound): {lower_bound:.2f}")
    print(f"Límite Superior (Upper Bound): {upper_bound:.2f}")
    print(f"Número de Outliers detectados: {len(outliers)}")
    if len(current_data) > 0:
        print(f"Porcentaje de Outliers: {(len(outliers) / len(current_data)) * 100:.2f}%")

    # Visualización de Boxplot para los outliers
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=current_data, color='#002D72')
    plt.title(f'Boxplot para {col} (Outliers detectados)')
    plt.xlabel(col)
    plt.show()

print("\n### Resumen General de Outliers por Variable ###")
display(outliers_summary)
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Manejar valores nulos para 'Arrival Delay in Minutes'
# Se identificó que 'Arrival Delay in Minutes' tiene 310 nulos y está muy asimétrica.
# Se imputa con la mediana para robustez contra asimetría y outliers.
if 'Arrival Delay in Minutes' in df.columns:
    median_arrival_delay = df['Arrival Delay in Minutes'].median()
    df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(median_arrival_delay)
    print(f"Valores nulos en 'Arrival Delay in Minutes' imputados con la mediana: {median_arrival_delay:.2f}")
else:
    print("Advertencia: La columna 'Arrival Delay in Minutes' no se encontró en el DataFrame.")

# Definir columnas para cada tipo de transformación basándose en el análisis de asimetría (Fase III)
# y el requisito del usuario sobre 'pisto' vs 'conducta'.

# Columnas para Transformación Logarítmica (altamente asimétricas positivamente, representando magnitudes)
log_transform_columns = ['Flight Distance', 'Departure Delay in Minutes', 'Arrival Delay in Minutes']

# Columnas para Normalización Z-Score (otras variables cuantitativas, incluyendo escalas de rating y edad)
# Excluir 'Unnamed: 0' e 'id' ya que son identificadores y no features para modelado.
# Excluir columnas ya seleccionadas para transformación logarítmica.
quantitative_cols_for_standardization = [
    col for col in df.select_dtypes(include=['int64', 'float64']).columns
    if col not in ['Unnamed: 0', 'id'] and col not in log_transform_columns
]

print(f"\nColumnas para Transformación Logarítmica (np.log1p): {log_transform_columns}")
print(f"Columnas para Normalización Z-Score: {quantitative_cols_for_standardization}")

# Aplicar Transformación Logarítmica (log1p)
for col in log_transform_columns:
    if col in df.columns:
        # np.log1p(x) calcula log(1+x) - útil para datos con ceros o valores pequeños
        df[col] = np.log1p(df[col])
        print(f"Columna '{col}' transformada con np.log1p().")
    else:
        print(f"Advertencia: La columna '{col}' no se encontró en el DataFrame para la transformación logarítmica.")

# Aplicar Normalización Z-Score
if quantitative_cols_for_standardization:
    scaler = StandardScaler()
    # Ajustar y transformar las columnas seleccionadas
    df[quantitative_cols_for_standardization] = scaler.fit_transform(df[quantitative_cols_for_standardization])
    print(f"\nColumnas {quantitative_cols_for_standardization} normalizadas con Z-Score.")
else:
    print("\nNo hay columnas cuantitativas restantes para aplicar Z-Score Normalization.")

print("\nDataFrame después de las transformaciones de normalización/escalado:")
print(df.head())
print("\nInformación del DataFrame actualizado:")
df.info()

# Opcional: Re-evaluar asimetría para columnas transformadas (muestras) para verificar la efectividad
print("\nSesgo de las columnas transformadas (muestras) para verificar la reducción de asimetría:")
for col in log_transform_columns:
    if col in df.columns:
        print(f"Sesgo de '{col}' después de log1p: {df[col].skew():.4f}")
for col in quantitative_cols_for_standardization:
    if col in df.columns:
        print(f"Sesgo de '{col}' después de Z-Score: {df[col].skew():.4f}")
```

```python
# Importar la función necesaria para la división de datos
from sklearn.model_selection import train_test_split
from pandas.api.types import is_numeric_dtype

# Paso 1: Convertir la variable 'satisfaction' a formato numérico
# Identificamos las clases en 'satisfaction' para entender su distribución.
print("Distribución original de la variable 'satisfaction':")
print(df['satisfaction'].value_counts())

# Mapear 'satisfied' a 1 y 'neutral or dissatisfied' a 0.
# 'satisfied' es la clase minoritaria en este dataset.
# Si 'Churn' se refiere a 'neutral or dissatisfied', debemos ser conscientes de que es la clase mayoritaria.
# La estratificación garantizará que la proporción de AMBAS clases se mantenga.
if not is_numeric_dtype(df['satisfaction']):
    df['satisfaction'] = df['satisfaction'].map({'satisfied': 1, 'neutral or dissatisfied': 0})

print("\nDistribución numérica de la variable 'satisfaction' después del mapeo:")
print(df['satisfaction'].value_counts())

# Paso 2: Definir las variables predictoras (X) y la variable objetivo (y)
# Excluir 'Unnamed: 0' e 'id' ya que son identificadores y no características para el modelo.
X = df.drop(columns=['satisfaction', 'Unnamed: 0', 'id'])
y = df['satisfaction']

# Asegurar que los modelos no reciban valores faltantes.
missing_values = X.isna().sum()
missing_values = missing_values[missing_values > 0]
if not missing_values.empty:
    print("\nValores faltantes detectados en X; se imputar?n con la mediana:")
    print(missing_values)
    X = X.fillna(X.median(numeric_only=True))

# Paso 3: Realizar la partición de datos con muestreo estratificado
# test_size=0.2 indica un 20% para el conjunto de evaluación
# stratify=y asegura que la proporción de clases en 'y' se mantenga en X_train, X_test, y_train, y_test
# random_state asegura la reproducibilidad de la partición
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(f"\nForma de X_train (conjunto de entrenamiento de características): {X_train.shape}")
print(f"Forma de X_test (conjunto de evaluación de características): {X_test.shape}")
print(f"Forma de y_train (conjunto de entrenamiento de etiquetas): {y_train.shape}")
print(f"Forma de y_test (conjunto de evaluación de etiquetas): {y_test.shape}")

print("\nVerificación de la distribución de clases en los conjuntos:")
print("Distribución de clases en y_train (entrenamiento):")
print(y_train.value_counts(normalize=True))

print("Distribución de clases en y_test (evaluación):")
print(y_test.value_counts(normalize=True))

print("\n¡La partición de los datos se ha realizado exitosamente con muestreo estratificado!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
import lightgbm as lgb
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB # Añadir Gaussian Naive Bayes
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import time
from pandas.api.types import is_numeric_dtype


# Asegurar que las variables de entrenamiento est?n limpias aunque las celdas previas
# se hayan ejecutado en otro orden.
if not is_numeric_dtype(y_train):
    y_train = y_train.map({'satisfied': 1, 'neutral or dissatisfied': 0})
if not is_numeric_dtype(y_test):
    y_test = y_test.map({'satisfied': 1, 'neutral or dissatisfied': 0})

missing_train = X_train.isna().sum()
missing_train = missing_train[missing_train > 0]
if not missing_train.empty:
    print("Valores faltantes detectados en X_train; se imputar?n con la mediana:")
    print(missing_train)
    train_medians = X_train.median(numeric_only=True)
    X_train = X_train.fillna(train_medians)
    X_test = X_test.fillna(train_medians)

# Diccionario para almacenar los modelos y sus resultados
models = {
    'Logistic Regression': LogisticRegression(random_state=42, solver='liblinear'),
    'Random Forest': RandomForestClassifier(random_state=42, n_jobs=-1),
    'LightGBM': lgb.LGBMClassifier(random_state=42, n_jobs=-1),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=5, n_jobs=-1),
    'Naive Bayes': GaussianNB() # Añadir Naive Bayes
}

# Diccionario para almacenar las métricas de cada modelo
results = {}

print("Iniciando el entrenamiento y evaluación de los modelos...")

for name, model in models.items():
    start_time = time.time()
    print(f"\nEntrenando {name}...")
    try:
        model.fit(X_train, y_train)
        training_time = time.time() - start_time
        print(f"{name} entrenado en {training_time:.2f} segundos.")

        # Predicciones en el conjunto de prueba
        y_pred = model.predict(X_test)

        # Calcular métricas
        accuracy = accuracy_score(y_test, y_pred)
        positive_label = 1 if is_numeric_dtype(y_test) else 'satisfied'
        precision = precision_score(y_test, y_pred, pos_label=positive_label)
        recall = recall_score(y_test, y_pred, pos_label=positive_label)
        f1 = f1_score(y_test, y_pred, pos_label=positive_label)

        results[name] = {
            'Accuracy': accuracy,
            'Precision': precision,
            'Recall': recall,
            'F1-Score': f1,
            'Training Time': training_time
        }

        print(f"--- Métricas para {name} ---")
        print(f"Accuracy: {accuracy:.4f}")
        print(f"Precision: {precision:.4f}")
        print(f"Recall: {recall:.4f}")
        print(f"F1-Score: {f1:.4f}")
    except Exception as e:
        print(f"Error al entrenar o evaluar {name}: {e}")
        results[name] = {"Error": str(e)}

print("\n--- Resumen de Resultados de Modelos ---")
for name, metrics in results.items():
    print(f"\nModelo: {name}")
    if "Error" in metrics:
        print(f"  Error: {metrics['Error']}")
    else:
        for metric_name, value in metrics.items():
            if "Time" in metric_name:
                print(f"  {metric_name}: {value:.2f} segundos")
            else:
                print(f"  {metric_name}: {value:.4f}")
```

### El Tribunal de Métricas: Presentación y Comparación de Modelos

En este análisis, hemos evaluado varios modelos de clasificación para predecir la satisfacción del cliente. La elección del "mejor" modelo no siempre se basa únicamente en la precisión general, sino en qué métrica es más crítica para los objetivos del negocio. A continuación, se explican las métricas clave y se fundamenta la elección del modelo ganador.

#### **1. Definición de las Métricas de Evaluación**

*   **Accuracy (Exactitud)**:
    *   **Definición**: Es la proporción de predicciones correctas (tanto verdaderos positivos como verdaderos negativos) sobre el número total de casos. Representa la capacidad general del modelo para clasificar correctamente.
    *   **Importancia**: Es útil cuando las clases están balanceadas y todos los errores de clasificación tienen el mismo costo. Sin embargo, puede ser engañosa en datasets desbalanceados.

*   **Precision (Precisión)**:
    *   **Definición**: De todas las predicciones positivas realizadas por el modelo, la precisión mide la proporción de esas predicciones que fueron realmente correctas (verdaderos positivos). Es decir, ¿cuántos de los clientes que el modelo dijo que estaban 'satisfechos' realmente lo estaban?
    *   **Importancia**: Es crucial cuando el costo de un falso positivo (clasificar erróneamente un cliente como satisfecho cuando no lo está) es alto. Por ejemplo, si se implementa una campaña de fidelización costosa solo para clientes "satisfechos", una alta precisión asegura que los recursos se invierten eficientemente.

*   **Recall (Sensibilidad o Exhaustividad)**:
    *   **Definición**: De todos los casos positivos reales, el Recall mide la proporción de los que fueron correctamente identificados por el modelo. Es decir, ¿cuántos de los clientes que *realmente* estaban 'satisfechos' fueron capturados por el modelo?
    *   **Importancia**: Es fundamental cuando el costo de un falso negativo (no identificar un cliente satisfecho cuando sí lo está) es alto. En el contexto de satisfacción del cliente, si el objetivo es asegurar que ningún cliente satisfecho pase desapercibido (por ejemplo, para ofrecerles promociones especiales o para un análisis más profundo de sus patrones de comportamiento positivo), un alto Recall es vital.

*   **F1-Score (Puntuación F1)**:
    *   **Definición**: Es la media armónica de Precision y Recall. Proporciona un equilibrio entre ambas métricas.
    *   **Importancia**: Es útil cuando se busca un balance entre Precision y Recall, especialmente en clases desbalanceadas, y cuando tanto los falsos positivos como los falsos negativos son importantes.

#### **2. Fundamentación del Modelo Ganador: Importancia del Recall en este Proyecto**

Dado el contexto de analizar la satisfacción del cliente, la métrica **Recall (Sensibilidad)** es fundamental. Si nuestro objetivo es comprender y capitalizar a los clientes *satisfechos*, o por el contrario, identificar a los *insatisfechos* para aplicar estrategias de retención, el costo de un **falso negativo** puede ser alto:

*   **Si la clase positiva es 'satisfecho'**: Un falso negativo significaría que un cliente realmente satisfecho es clasificado erróneamente como 'neutral o insatisfecho'. Esto podría llevar a que se pierdan oportunidades de fortalecer la lealtad de esos clientes, dejándolos fuera de programas de recompensa o segmentaciones de marketing específicas para clientes de alto valor.
*   **Si la clase positiva es 'insatisfecho' (o 'churn')**: Un falso negativo implicaría no identificar a un cliente en riesgo de insatisfacción o abandono. Esto es crítico, ya que la detección temprana es clave para intervenir y potencialmente retener a esos clientes. En este proyecto específico, nuestra variable objetivo mapeó 'satisfied' a 1, por lo que nos enfocamos en identificar correctamente a los satisfechos.

Por lo tanto, un modelo con alto Recall es crucial para **maximizar la detección de la clase de interés**, minimizando los casos perdidos.

### Fórmulas Financieras Utilizadas en la Auditoría

#### 1. Revenue at Risk (Flujo Mensual Expuesto)
Esta métrica cuantifica cuánto dinero dejaría de percibir la empresa el próximo mes si no se interviene a los clientes identificados por el modelo.

$$\text{Revenue at Risk} = \text{Clientes en Riesgo} \times \text{Ticket Promedio Mensual}$$

*Donde 'Clientes en Riesgo' son las predicciones del modelo para la clase 0 (Neutral o Insatisfecho).*

#### 2. Lifetime Value (LTV) Proyectado
Calculamos el valor total que un cliente aporta a la compañía durante toda su relación comercial. La fórmula de perpetuidad es:

$$\text{LTV} = \frac{\text{Ingreso Mensual}}{\text{Tasa de Fuga Mensual (Churn Rate)}}$$

Para obtener la **Tasa de Fuga Mensual** a partir de la Retención Anual:
1. $\text{Tasa de Fuga Anual} = 1 - \text{Tasa de Retención Anual}$
2. $\text{Tasa de Fuga Mensual} = \frac{\text{Tasa de Fuga Anual}}{12}$

```python
# Resumen de la lógica aplicada en código:
ticket = 500
retencion_anual = 0.85
fuga_anual = 1 - retencion_anual  # 0.15
fuga_mensual = fuga_anual / 12    # 0.0125 o 1.25%

ltv = ticket / fuga_mensual

print(f"Fórmula LTV: {ticket} / {fuga_mensual} = {ltv}")
```
